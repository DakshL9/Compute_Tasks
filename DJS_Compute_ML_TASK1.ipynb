**Question 1: Coded Warnings (Strings)**

Peter can't risk MJ or Ned figuring out who's really behind these
messages if a text gets intercepted, so he's adopted a simple code:
every alternate word in the message is reversed. The first word stays
as-is, the second is reversed, the third stays as-is, and so on.

Write a function `encrypt_message(message)` that takes a sentence and
returns the encoded version.

Example:
Input:  "Stay inside tonight"
Output: "Stay edisni tonight"

```python
message = "Meet me at the bridge tonight before curfew"
```

```python
# your code here
# Expected Output:
# 'Meet em at eht bridge thginot before wefruc'

msg="Meet me at the bridge tonight before curfew"

words=msg.split()
msg=""
for i in range(0,len(words)):
  if i%2!=0:
    words[i]=words[i][::-1]
  msg=msg+ " " + words[i]
print(msg.strip())
```
